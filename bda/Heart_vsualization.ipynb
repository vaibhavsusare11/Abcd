```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv("heart11.csv")
```

```python
df
```

```python
df.isna().sum()
```

```python
#heatmap
#correlation between numeric columns ranges from -1 to 1
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(),cmap="YlGnBu",annot=True)
```

```python
#line plot
#max heart rate by age
sns.lineplot(data=df.sort_values(by="age"),x='age',y='thalachh',marker='o',color="blue")
```

```python
#high cholestrol risk with the age
sns.lineplot(data=df.sort_values(by="age"),x="age",y="chol",marker="o",color="red")
```

```python
#age vs blood pressure
sns.lineplot(data=df.sort_values(by="age"),x="age",y="trtbps",marker="o",color="green")
```

```python
#barplot
#differt type of chest pain vs cholestrol level
sns.barplot(data=df,x='cp',y='chol',palette='coolwarm')
```

```python
#countplot
#each type of chest pain how many patients
sns.countplot(data=df,x='cp',palette='coolwarm')
```

```python
#histogram
#age distribution with heart disease
sns.histplot(data=df,x='age',hue='output',kde=True)
```

```python
sns.scatterplot(data=df,x="age",y="thalachh",hue="output")
```

```python
sns.violinplot(data=df,y='chol',inner='point')
plt.title("Cholestrol level distribution")
plt.ylabel("cholestrol level")
plt.show()
```

```python
sns.stripplot(data=df,x="output",y="oldpeak",hue="output",palette="bright",jitter=True)
```

```python
sns.boxplot(data=df, x="output", y="thalachh", palette="bright")
```

```python
from sklearn.cluster import KMeans
X = df[['age', 'thalachh']]
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X)
```

```python
sns.scatterplot(data=df, x="age", y="thalachh", hue="cluster", style="cluster")
```
