```python
import pandas as  pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("AirQuality_visualization.csv",delimiter=';')
```

```python
df.head()
```

```python
df=df.rename(columns={'T':'Temperature'})
df=df.rename(columns={'RH':'Relative Humidity'})
df=df.rename(columns={'AH':'Absolute Humidity'})
df
```

```python
df = df.drop(['Unnamed: 15', 'Unnamed: 16'], axis=1, errors='ignore')
```

```python
df
```

```python
df = df.replace(',', '.', regex=True)
```

```python
df
```

```python
float_cols = ['CO(GT)', 'C6H6(GT)', 'Temperature', 'Relative Humidity', 'Absolute Humidity']
df[float_cols] = df[float_cols].astype(float)
```

```python
df=df.drop_duplicates()
```

```python
df.isna().sum()
```

```python
df=df.fillna(df.mean(numeric_only=True))
df=df.dropna()
```

```python
df.isna().sum()
```

```python
df['Absolute Humidity']=df['Absolute Humidity'].multiply(100)
```

```python
def remove_outliers(column):
    Q1=column.quantile(0.25)
    Q3=column.quantile(0.75)
    IQR=Q3-Q1
    threshold=1.5*IQR
    outlier_mask=(column<Q1-threshold) | (column>Q3+threshold)
    return column[~outlier_mask]
```

```python
col_name = ['Temperature', 'Relative Humidity', 'Absolute Humidity', 
            'PT08.S4(NO2)', 'PT08.S5(O3)', 'C6H6(GT)', 
            'PT08.S2(NMHC)', 'PT08.S1(CO)']

for col in col_name:
    df[col] = remove_outliers(df[col])
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y", errors='coerce')
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
```

```python
df['Year'] = df.Year.astype(str)
df['Month'] = df.Month.astype(str)
```

```python
df
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap='coolwarm')
```

```python
sns.lineplot(data=df,x='Month',y="Absolute Humidity")
```

```python
sns.lineplot(data=df,x="Month",y="Temperature")
```

```python
sns.barplot(data=df, x="Month", y='CO(GT)')
```

```python
sns.scatterplot(df, x='PT08.S1(CO)', y='PT08.S5(O3)', hue='Year')
```

```python
#	Get	the	top	5	most	frequent	CO	values
co_counts=df["CO(GT)"].value_counts().nlargest(5)
 #	Simple	Pie	Chart
plt.figure(figsize=(6,	6))
co_counts.plot.pie(
    autopct="%1.1f%%",		
    startangle=90		
)		
plt.title("CO Levels Distribution")		
plt.ylabel("")		#	Hide	default	y-label		
plt.show()
```

```python
sns.histplot(df["Temperature"], kde=True, color="red")
```

```python
df["Category"]="CO Levels"
sns.violinplot(data=df,x="Category",y="CO(GT)",color="skyblue")		
plt.title("CO	Levels	Violin	Plot")		
plt.xlabel("Category")		
plt.ylabel("CO(GT)")		
plt.show()
```

```python
sns.boxplot(data=df,x='Month',y='Temperature')
plt.title('Box	Plot of	Temperature	by	Month')
plt.xlabel('Month')
plt.ylabel('Temperature')
plt.show() 
```
